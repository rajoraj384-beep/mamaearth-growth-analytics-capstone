```python
import pandas as pd
import numpy as np

# Step 1: Teeno CSV files ko load karein
customers = pd.read_csv("customers.csv")
products = pd.read_csv("products.csv")
orders = pd.read_csv("orders.csv")

print("--- Data Loaded Successfully ---")
print("Customers shape:", customers.shape)
print("Products shape:", products.shape)
print("Orders shape:", orders.shape)

# Step 2: Payment methods ko standardise karein (Uppercase & Strip)
orders["payment_method"] = orders["payment_method"].str.strip().str.upper()
print("\nPayment methods breakdown:\n", orders["payment_method"].value_counts())

# Step 3: Duplicate orders remove karein
natural_keys = [
    "customer_id",
    "product_id",
    "order_date",
    "quantity",
    "discount_pct",
    "payment_method",
    "rating",
    "returned"
]
orders_clean = orders.drop_duplicates(subset=natural_keys, keep="first").copy()
print(f"\nOriginal orders count: {len(orders)}")
print(f"Cleaned orders count after removing duplicates: {len(orders_clean)}")
```

```python
# Step 4: Missing values ko impute karna
orders_clean["discount_pct"] = orders_clean["discount_pct"].fillna(0)

# Rating ke missing values ko median rating se fill karein
median_rating = orders_clean["rating"].median()
orders_clean["rating"] = orders_clean["rating"].fillna(median_rating)

print("Missing values after imputation:\n", orders_clean.isnull().sum())
```

```python
# Step 5: Tables ko merge karein aur Net Revenue calculate karein
merged = orders_clean.merge(products, on="product_id").merge(customers, on="customer_id")

# Net order value formula: quantity * price * (1 - discount_pct / 100)
merged["order_value"] = merged["quantity"] * merged["price"] * (1 - merged["discount_pct"] / 100.0)

total_revenue = round(merged["order_value"].sum(), 2)
print(f"Cleaned Total Revenue (INR): {total_revenue}")
print("Merged dataset shape:", merged.shape)
```

```python
# Step 6: Outlier Detection using IQR method on Quantity
Q1 = orders_clean["quantity"].quantile(0.25)
Q3 = orders_clean["quantity"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = orders_clean[(orders_clean["quantity"] < lower_bound) | (orders_clean["quantity"] > upper_bound)]
print(f"IQR range: {lower_bound} to {upper_bound}")
print(f"Number of quantity outliers detected: {len(outliers)}")
```

```python
# Step 7: Key Insights & Aggregations
print("--- Category-wise Revenue ---")
cat_revenue = merged.groupby("category")["order_value"].sum().reset_index()
print(cat_revenue)

print("\n--- Return Rate by Payment Method ---")
return_rate = merged.groupby("payment_method")["returned"].mean().reset_index()
return_rate["return_rate_pct"] = round(return_rate["returned"] * 100, 2)
print(return_rate[["payment_method", "return_rate_pct"]])
```
